```python
import streamlit as st
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

def main():
    st.title("📄 Text Summarizer & Question Answering")
    st.write("Upload or paste text, get a summary, and ask questions.")

    text_input = st.text_area("Enter your text here:", height=300)

    if st.button("Summarize"):
        if text_input.strip():
            with st.spinner("Summarizing..."):
                summary = summarizer(text_input, max_length=150, min_length=50, do_sample=False)[0]['summary_text']
            st.subheader("Summary:")
            st.write(summary)
        else:
            st.warning("Please enter some text to summarize.")

    st.subheader("Ask Questions:")
    question = st.text_input("Enter your question about the text:")
    if st.button("Get Answer"):
        if question.strip() and text_input.strip():
            with st.spinner("Finding answer..."):
                answer = qa_pipeline(question=question, context=text_input)
            st.write(f"**Answer:** {answer['answer']}")
        else:
            st.warning("Please provide both text and a question.")

if __name__ == "__main__":
    main()
```
