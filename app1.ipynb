```python
import streamlit as st
import numpy as np
import pickle
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
model = load_model('/content/drive/MyDrive/twitterdataset/sentiment_model.h5')
with open('/content/drive/MyDrive/twitterdataset/tokenizer.pkl', 'rb') as f:
    tokenizer = pickle.load(f)

with open('/content/drive/MyDrive/twitterdataset/label_map.pkl', 'rb') as f:
    label_map = pickle.load(f)

max_len = 100
def clean_text(text):
    text = str(text)
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = text.lower()
    text = " ".join([word for word in text.split() if word not in stop_words])
    return text
    st.title("Twitter Sentiment Analysis")
st.write("Predict sentiment of a tweet using your trained LSTM model.")

user_input = st.text_area("Enter a tweet:")

if st.button("Analyze Sentiment"):
    if user_input.strip():
        cleaned = clean_text(user_input)
        seq = tokenizer.texts_to_sequences([cleaned])
        pad_seq = pad_sequences(seq, maxlen=max_len)

        pred = model.predict(pad_seq)
        predicted_label = label_map[np.argmax(pred)]

        st.success(f"Predicted Sentiment: {predicted_label}")
    else:
        st.warning("Please enter some text.")
        st.title("Twitter Sentiment Analysis")

```
